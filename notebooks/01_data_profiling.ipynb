**DATA CLEANING**

```python
cleaned_df = df.copy()
```

```python
cleaned_df["Parent_Education_Level"] = cleaned_df["Parent_Education_Level"].fillna("Unknown")
```

```python
cleaned_df["Parent_Education_Level"].isnull().sum()
```

```python
df["Parent_Education_Level"].isnull().sum()
```

```python
cleaned_df.columns.tolist()
```

```python
pii_columns = ["First_Name","Last_Name","Email"]
cleaned_df = cleaned_df.drop(columns=pii_columns)
```

```python
cleaned_df.columns.tolist()
```

```python
cleaned_df.shape
```

```python
cleaned_df["Total_Score"].describe()
```

```python
cleaned_df.groupby("Grade")["Total_Score"].agg(["count", "min", "max", "mean"])
```

```python
cleaned_df.groupby("Grade")["Attendance (%)"].agg(["count", "mean", "min", "max"])
```

```python
assessment_columns = [
    "Midterm_Score",
    "Final_Score",
    "Assignments_Avg",
    "Quizzes_Avg",
    "Participation_Score",
    "Projects_Score"
]

cleaned_df["Average_Assessment_Score"] = cleaned_df[assessment_columns].mean(axis=1)
```
