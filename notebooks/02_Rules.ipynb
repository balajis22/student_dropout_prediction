Data Quality rules

Student ID
 - must not be missing
 - Must be uniqe

Attendance (%)
 - Must be between 0 and 100.

Score columns
 - Midterm_Score
 - Final_Score
 - Assignments_Avg
 - Quizzes_Avg
 - Participation_Score
 - Projects_Score
 - Total_Score
 - All should be between 0 and 100.

Age
 - Should be within a reasonable student-age range.
 - We'll flag values outside 18–25, since that's the observed range in this dataset

Categorical fields
 - Gender, Department, Family Income, etc. should contain only expected categories.

Parent_Education_Level
 - Missing values are allowed.
 - We won't delete those students just because this field is missing.
 - Later we'll document how we handle them.

PII fields
 - First_Name, Last_Name, and Email won't be carried into the analytical layer.
